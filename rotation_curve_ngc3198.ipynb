```python
cols = ["Rad", "Vobs", "errV", "Vgas", "Vdisk", "Vbul", "SBdisk", "SBbul"] # comment :D always comment !!!!
indir = "/Users/silvana/Desktop/mentoria_proyecto/project1/Rotmod_LTG/"

gal = pd.read_csv(indir + "NGC3109_rotmod.dat",comment="#", delim_whitespace=True, names=cols)
```

```python
gal
```

```python
r_kpc = gal['Rad'] #radius in kiloparsec 
vobs = gal['Vobs'] # observed velocity in km/s 
verr = gal['errV'] # error in the observed velocity
```

```python
plt.figure(figsize=(9,6))

plt.errorbar(r_kpc, vobs, lw=4,color='red',yerr=verr,elinewidth=2,fmt='^',markersize=10,label='Observed')

plt.plot(r_kpc, gal["Vgas"],lw=3, color='blue',label="Gas")
plt.plot(r_kpc, gal["Vdisk"],lw=3,color='lightseagreen',label="Disk")
plt.plot(r_kpc, gal["Vbul"], color='violet', lw=3,label="Bulge")

plt.xlabel('Radius [kpc]', fontsize=20)
plt.ylabel('Velocity [km/s]', fontsize=20)

plt.title('NGC 3109',fontsize=25)

plt.legend(fontsize=18)
plt.ylim([0,80])


plt.show()
```
