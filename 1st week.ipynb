```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('NGC3109_rotmod.dat',sep=r"\s+", names=["Rad", "Vobs","errV","Vgas","Vdisk","Vbul","SBdisk","SBbu"],comment="#")
print(df) 
```

```python
x=df["Rad"]
```

```python
fig,ax=plt.subplots()
ax.plot(x,df["Vgas"],label="Gas",color="blue")
ax.plot(x,df["Vdisk"],label="Disk",color="cyan")
ax.plot(x,df["Vbul"],label="Bulge",color="pink")
ax.errorbar(x,df["Vobs"], yerr=df["errV"], fmt='o',color="red",label="Observed",elinewidth=1)
plt.xlabel("Radius (kpc)")
plt.ylabel("Velocity (km/s)")
plt.title("NGC3109")
plt.legend(fontsize=13)
ax.set_ylim(0,80)
plt.show()
```
