```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
cols=["Galaxy","Distance (Mpc)","MBH","MBH lower","MBH upper","method","sigma","sigma lower",
                                          "sigma upper","log(LV/Lsun)","error in log(LV/Lsun)","Mbulge/Msun","radius of influence (arcsec)",
                                          "Morphology","Profile"]
```

```python
df=pd.read_csv("McConnell_Ma_2013_ascii.txt",sep=r"\s+",names=cols,skiprows=17)
```

```python
df.head
```

```python
#A few new columns was created 
df["yerr_low"] = df["MBH"] - df['MBH lower']
df["yerr_up"] = df['MBH upper'] - df['MBH']

df["xerr_low"] = df['sigma'] - df['sigma lower']
df["xerr_up"] = df['sigma upper'] - df['sigma']
```

```python
fig,ax=plt.subplots()
ax.scatter(df["sigma"],df["MBH"],color="blue",label="Galaxies",s=20)
ax.errorbar(df["sigma"], df["MBH"], xerr=[df["xerr_low"],df["xerr_up"]],yerr=[df["yerr_low"],df["yerr_up"]] , fmt=' ',color="black",
            elinewidth=1,capsize=1)

logx=np.log10(df["sigma"]/200.00)
logy=np.log10(df["MBH"])
coefs=np.polyfit(logx,logy,1)
x=np.linspace(logx.min(),logx.max(),100)
y=coefs[0]*x+coefs[1]
ax.plot(200.0*10**x,10**y,linestyle="--",c="b")

plt.ylabel("Black Hole Masses $M_{BH}$")
plt.xlabel("Stellar Velocity Dispersion $\sigma$ (km/s)")
plt.xscale("log")
plt.yscale("log")
plt.title("$M_{BH}$ vs $\sigma$")
plt.legend()
for i, row in df.iterrows():
    plt.annotate(
        row["Galaxy"],
        (row["sigma"], row["MBH"]),
        xytext=(3, 3),
        textcoords="offset points",
        fontsize=6)
plt.show()
```

Plot a)
For this plot, I did three new dataframes in order to separate the information in groups. 


```python
df_S=df[df["method"]=="star"]
df_G=df[df["method"]=="gas"]
df_M=df[df["method"]=="maser"]
```

```python
#A linear regresion for each group
logxs=np.log10(df_S["sigma"]/200.00)
logys=np.log10(df_S["MBH"])
A,B =np.polyfit(logxs,logys,1)
x_s=np.linspace(logxs.min(),logxs.max(),100)
y_s=A*x_s+B
```

```python
logxg=np.log10(df_G["sigma"]/200)
logyg=np.log10(df_G["MBH"])
C,D =np.polyfit(logxg,logyg,1)
x_g=np.linspace(logxg.min(),logxg.max(),100)
y_g=C*x_g+D
```

```python
logxm=np.log10(df_M["sigma"]/200)
logym=np.log10(df_M["MBH"])
E,F =np.polyfit(logxm,logym,1)
x_m=np.linspace(logxm.min(),logxm.max(),100)
y_m=E*x_m+F
```

```python
fig,ax=plt.subplots()

ax.errorbar(df["sigma"], df["MBH"], xerr=[df["xerr_low"],df["xerr_up"]],yerr=[df["yerr_low"],df["yerr_up"]] , fmt=' ',color="black",
            elinewidth=1,capsize=1,zorder=1)
ax.scatter(df_S["sigma"],df_S["MBH"],s=50,color="green",marker="*",label="Star")
ax.scatter(df_G["sigma"],df_G["MBH"],color="blue",marker=".",s=50,label="Gas")
ax.scatter(df_M["sigma"],df_M["MBH"],color="red",marker="^",label="Masser")
ax.plot(200.00*10**x_s,10**y_s,c="g",linestyle="--",zorder=2)
ax.plot(200*10**x_g,10**y_g,c="b",linestyle="--",zorder=2)
ax.plot(200*10**x_m,10**y_m,c="r",linestyle="--",zorder=2)
plt.ylabel("Black Hole Masses $M_{BH}$")
plt.xlabel("Stellar Velocity Dispersion $\sigma$ (km/s)")
plt.xscale("log")
plt.yscale("log")
plt.title("$M_{BH}$ vs $\sigma$")
plt.legend(title = "Method")

for i, row in df.iterrows():
    plt.annotate(
        row["Galaxy"],
        (row["sigma"], row["MBH"]),
        xytext=(3, 3),
        textcoords="offset points",
        fontsize=6)
plt.show()
```

```python
df_S=df[df["Morphology"]=="S"]
df_S0=df[df["Morphology"]=="S0"]
df_E=df[df["Morphology"]=="E"]
```

```python
logxs=np.log10(df_S["sigma"]/200)
logys=np.log10(df_S["MBH"])
G,H =np.polyfit(logxs,logys,1)
x_s=np.linspace(logxs.min(),logxs.max(),100)
y_s=G*x_s+H
```

```python
logxs0=np.log10(df_S0["sigma"]/200)
logys0=np.log10(df_S0["MBH"])
I,J =np.polyfit(logxs0,logys0,1)
x_s0=np.linspace(logxs0.min(),logxs0.max(),100)
y_s0=I*x_s0+J
```

```python
logxe=np.log10(df_E["sigma"]/200)
logye=np.log10(df_E["MBH"])
K,L =np.polyfit(logxe,logye,1)
x_e=np.linspace(logxe.min(),logxe.max(),100)
y_e=K*x_e+L
```

```python
fig,ax=plt.subplots()

ax.errorbar(df["sigma"], df["MBH"], xerr=[df["xerr_low"],df["xerr_up"]],yerr=[df["yerr_low"],df["yerr_up"]] , fmt=' ',color="black",
        elinewidth=1,capsize=1,zorder=1)
ax.scatter(df_S["sigma"],df_S["MBH"],s=50,color="green",marker="*",label="S")
ax.scatter(df_S0["sigma"],df_S0["MBH"],color="blue",marker=".",s=50,label="S0")
ax.scatter(df_E["sigma"],df_E["MBH"],color="red",marker="^",label="E")
ax.plot(200*10**x_s,10**y_s,c="g",linestyle="--",zorder=2)
ax.plot(200*10**x_s0,10**y_s0,c="b",linestyle="--",zorder=2)
ax.plot(200*10**x_e,10**y_e,c="r",linestyle="--",zorder=2)

plt.ylabel("Black Hole Masses $M_{BH}$")
plt.xlabel("Stellar Velocity Dispersion $\sigma$ (km/s)")
plt.xscale("log")
plt.yscale("log")
plt.title("$M_{BH}$ vs $\sigma$")
plt.legend(title = "Morphology")
for i, row in df.iterrows():
    plt.annotate(
        row["Galaxy"],
        (row["sigma"], row["MBH"]),
        xytext=(3, 3),
        textcoords="offset points",
        fontsize=6)
plt.show()
```
